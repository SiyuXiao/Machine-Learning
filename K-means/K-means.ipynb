```python
#刚开始随便选两个点，然后用所有点和这两个点算距离，然后更新这两个点不断迭代直到收敛
X = [1, 2, 3, 4, 8, 9, 10]

centroids = [1, 3]
new_centroids = []

while True:
    cluster1 = []
    cluster2 = []
    for x in X:
        if abs(x - centroids[0]) < abs(x - centroids[1]):
            cluster1.append(x)
        else:
            cluster2.append(x)
    new_centroids = [
        sum(cluster1) / len(cluster1),
        sum(cluster2) / len(cluster2)
    ]

    print(cluster1, cluster2)
    print(new_centroids)
    if centroids == new_centroids:
        break
    else:
        centroids = new_centroids
centroids
```
